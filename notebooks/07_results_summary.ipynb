# 07 Results summary

Reads only the saved JSON files in `outputs/results`, so every number here can be traced back. Writes `RESULTS.md` in the project folder. Fill the "what this means" text by hand after checking the numbers.

```python
%run ./00_common.ipynb
def load(name):
    p = RESULTS / name
    return json.load(open(p)) if p.exists() else None
s01, s02, s03, s04, s05, s06 = [load(n) for n in ['01_setup.json', '02_reproduce_1d_synthetic.json', '03_dataset.json',
                                                  '04_predictor.json', '05_warm_start_summary.json', '06_quantisation.json']]
def fmt(d): return ', '.join(f'{k} {v:.2f}' for k, v in d.items()) if isinstance(d, dict) else str(d)
lines = ['# Results: one shot and warm started inverse simulation for MIMO-SAR (synthetic data)', '',
         'All numbers are read from outputs/results/*.json. Synthetic data from the public InverseTHzSim simulator; real IEEE DataPort measurements not used yet.', '']
if s01: lines += [f"* Simulator: Mitsuba {s01['mitsuba_version']}, {s01['seconds_per_sim']:.2f} s per simulated measurement on the RTX 3090."]
if s02: lines += [f"* Paper method reproduced (1D x search, cold start, synthetic target): final error {s02['final_err_mm']:.2f} mm, reached 1 mm after {s02['iters_to_tol']} iterations, {s02['seconds']:.0f} s."]
if s03: lines += [f"* Dataset: {s03['samples']} synthetic spheres."]
if s04: lines += [f"* One shot predictor test MAE (mm): {fmt(s04['test_mae_mm'])}. Mean baseline: {fmt(s04['mean_baseline_mae_mm'])}. {s04['ms_per_prediction']:.2f} ms per prediction, {s04['n_params']} parameters."]
if s05: lines += [f"* Warm start vs cold start over {s05['n_cases']} test spheres: cold {fmt({k: v for k, v in s05['cold'].items() if isinstance(v, (int, float))})}, reached 1 mm {s05['cold']['reached_1mm']}; warm {fmt({k: v for k, v in s05['warm'].items() if isinstance(v, (int, float))})}, reached 1 mm {s05['warm']['reached_1mm']}."]
if s06: lines += [f"* 8 bit: FP32 {fmt(s06['fp32_test_mae_mm'])} | {s06['final_method']} {fmt(s06['final_int8_test_mae_mm'])} | weights {s06['fp32_weight_kb']:.0f} KB to {s06['int8_weight_kb']:.0f} KB."]
lines += ['', '## Honest limitations', '* Synthetic data only; the real 235 to 270 GHz measurements (IEEE DataPort, DOI 10.21227/8jb1-x962) are the next test.',
          '* Spheres only; torus and plate need meshes and are future work.', '* Fake quantisation: accuracy effect only, nothing run on hardware yet.',
          '* Small warm start test (see n_cases); one training seed.', '', '## Figures', '* outputs/figures/']
any_fake = any(isinstance(s, dict) and s.get('fake_mode') for s in (s03, s04, s06))
if any_fake: lines.insert(2, '**WARNING: some results were made in FAKE_MODE (random test data). They are NOT real results.**\n')
open(PROJECT / 'RESULTS.md', 'w').write('\n'.join(lines) + '\n'); print('\n'.join(lines))
```
